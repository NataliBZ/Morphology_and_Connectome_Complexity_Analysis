# Figure 3 Testing different pruning methods showed differences in network complexity.
### A) Edge survival probability for the CPCs (solid lines) and the other pruned networks (dashed lines) for rat (right) and human (left).

We plot the probablity that an edge will survive a given prunning method given its number of touches

```python
from pathlib import Path
import pickle
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data 
fname="survival_probabilites.pkl"
with open(fname, "rb") as f:
    survival_prob = pickle.load(f)
```

```python
DF = survival_prob["rat"]
df_EtoT = pd.DataFrame.from_dict(DF["E_to_T"])
df_TtoT = pd.DataFrame.from_dict(DF["T_to_T"])
df_EtoE = pd.DataFrame.from_dict(DF["E_to_E"])
df_TtoE = pd.DataFrame.from_dict(DF["T_to_E"])
```

```python
mean_EtoT = df_EtoT.mean(axis=1)
sem_EtoT = df_EtoT.sem(axis=1)
mean_TtoT = df_TtoT.mean(axis=1)
sem_TtoT = df_TtoT.sem(axis=1)
mean_EtoE = df_EtoE.mean(axis=1)
sem_EtoE = df_EtoE.sem(axis=1)
mean_TtoE = df_TtoE.mean(axis=1)
sem_TtoE = df_TtoE.sem(axis=1)
```

```python
plt.figure()
plt.plot(survival_prob["rat"]["functional"], color='blueviolet', linestyle ='solid', label= 'CPC')
plt.plot(mean_EtoT, color='brown', linestyle ='dashed', label= 'E_to_T')
plt.fill_between(df_EtoT.index, mean_EtoT-sem_EtoT, mean_EtoT+sem_EtoT, alpha=0.3, color='brown')

plt.plot(df_TtoT.mean(axis=1), color = 'coral', linestyle ='dotted', label= 'T_to_T')
plt.fill_between(df_TtoT.index, mean_TtoT-sem_TtoT, mean_TtoT+sem_TtoT, alpha=0.3, color='coral')

plt.plot(df_EtoE.mean(axis=1), color = '#388004', linestyle ='dashed', label= 'E_to_E')
plt.fill_between(df_EtoE.index, mean_EtoE-sem_EtoE, mean_EtoE+sem_EtoE, alpha=0.3, color='#388004')

plt.plot(df_TtoE.mean(axis=1), color = '#35530a', linestyle ='dotted', label= 'T_to_E')
plt.fill_between(df_TtoE.index, mean_TtoE-sem_TtoE, mean_TtoE+sem_TtoE, alpha=0.3, color='#35530a')

plt.legend()

plt.title('Survival probability of touches rat')
plt.ylabel('Survival probability', fontsize=15)
plt.xlabel('Touch count', fontsize=15)

plt.xlim(0, 8)
plt.ylim(0, 0.8)
```

```python
DF = survival_prob["human"]
df_EtoT = pd.DataFrame.from_dict(DF["E_to_T"])
df_TtoT = pd.DataFrame.from_dict(DF["T_to_T"])
df_EtoE = pd.DataFrame.from_dict(DF["E_to_E"])
df_TtoE = pd.DataFrame.from_dict(DF["T_to_E"])
```

```python
mean_EtoT = df_EtoT.mean(axis=1)
sem_EtoT = df_EtoT.sem(axis=1)
mean_TtoT = df_TtoT.mean(axis=1)
sem_TtoT = df_TtoT.sem(axis=1)
mean_EtoE = df_EtoE.mean(axis=1)
sem_EtoE = df_EtoE.sem(axis=1)
mean_TtoE = df_TtoE.mean(axis=1)
sem_TtoE = df_TtoE.sem(axis=1)
```

```python
plt.figure()
plt.plot(survival_prob["human"]["functional"], color='teal', linestyle ='solid', label= 'CPC')
plt.plot(mean_EtoT, color='brown', linestyle ='dashed', label= 'Human: E_to_T')
plt.fill_between(df_EtoT.index, mean_EtoT-sem_EtoT, mean_EtoT+sem_EtoT, alpha=0.3, color='brown')

plt.plot(df_TtoT.mean(axis=1), color = 'coral', linestyle ='dotted', label= 'T_to_T')
plt.fill_between(df_TtoT.index, mean_TtoT-sem_TtoT, mean_TtoT+sem_TtoT, alpha=0.3, color='coral')

plt.plot(df_EtoE.mean(axis=1), color = '#388004', linestyle ='dashed', label= 'E_to_E')
plt.fill_between(df_EtoE.index, mean_EtoE-sem_EtoE, mean_EtoE+sem_EtoE, alpha=0.3, color='#388004')

plt.plot(df_TtoE.mean(axis=1), color = '#35530a', linestyle ='dotted', label= 'T_to_E')
plt.fill_between(df_TtoE.index, mean_TtoE-sem_TtoE, mean_TtoE+sem_TtoE, alpha=0.3, color='#35530a')

plt.legend()

plt.title('Survival probability of touches human')
plt.ylabel('Survival probability', fontsize=15)
plt.xlabel('Touch count', fontsize=15)

plt.xlim(0, 8)
plt.ylim(0, 0.8)
```
