# Figure S2 Touches, edges, and touches-per-edge across species and pruning types. 
#### Rates of number of touches over the number of edges for each network for human (teal) and rat (purple).
#### Dashed lines show mean value references from literature and shadows are the standard deviations respectively (top). 
#### Total number of touches (red) and edges (blue) in each built network, for the human and rat

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
circuit_names = ['ATC', 'CPC', 'E_to_T', 'T_to_T', 'E_to_E', 'T_to_E']
num_edges_human = [36785128, 3595201, 6178971, 12033027, 3595201, 3621764]
num_touches_human = [95266220, 16002313, 15991403, 16002313, 9309663, 3938934]
ratio_touches_edges_human = [2.59, 4.45, 2.59, 1.33, 2.6, 1.09]
num_edges_rat = [67124018, 3129285, 4206805, 9073536, 3129285, 3133728]
num_touches_rat = [160450072, 10055745, 10055886, 10055745, 7480926, 3246053]
ratio_touches_edges_rat = [2.39, 3.2, 2.39, 1.10, 2.39, 1.03]
```

```python
#fig = plt.figure(figsize=(6, 2.78))
fig = plt.figure(figsize=(8, 4))
ax0=plt.subplot()


ind = np.arange(len(circuit_names))
width = 0.45

s1 = ax0.bar(ind, num_touches_human, width, color='red', label='Touches')
s2 = ax0.bar(ind + width, num_edges_human, width, color='blue',label='Edges')

#ax0.set_xticks(ind + width / 2, rotation=45)
ax0.set_xticks(range(len(circuit_names)), labels=circuit_names, fontsize=12,
              rotation=45, ha="right", rotation_mode="anchor")

#ax0.set_xticklabels(circuit_names, fontsize=12, ha="center") #
#ax0.set_yticklabels(yticks, fontsize=12, ha="right") #

ax0.set_title('Number of touches and edges for human', fontsize=12)
ax0.set_ylabel('Number', fontsize=12)
ax0.set_xlabel('Connectivity matrix', fontsize=12)
ax0.spines[['right', 'top']].set_visible(False)
ax0.legend(loc='center left', bbox_to_anchor=(1, 0.5))

#plt.savefig('Number_touches_edges_Human.svg', bbox_inches='tight', transparent=True, dpi=100)
#plt.show()
```

```python
fig = plt.figure(figsize=(8, 4))
ax0=plt.subplot()


ind = np.arange(len(circuit_names))
width = 0.45

s1 = ax0.bar(ind, num_touches_rat, width, color='red', label='Touches')
s2 = ax0.bar(ind + width, num_edges_rat, width, color='blue',label='Edges')

ax0.set_xticks(range(len(circuit_names)), labels=circuit_names, fontsize=12,
              rotation=45, ha="right", rotation_mode="anchor")

ax0.set_title('Number of touches and edges for rat', fontsize=12)
ax0.set_ylabel('Number', fontsize=12)
ax0.set_xlabel('Connectivity matrix', fontsize=12)
ax0.spines[['right', 'top']].set_visible(False)
ax0.legend(loc='center left', bbox_to_anchor=(1, 0.5))

#plt.savefig('Number_touches_edges_Rat.svg', bbox_inches='tight', transparent=True, dpi=100)
#plt.show()
```

```python
fig = plt.figure(figsize=(8, 4))
ax0=plt.subplot()

# horizontal lines with experimental values:
# Human (Hunt et al., 2022): 4.0 ± 1.2 
# Rat (Feldmeyer et al., 2006): 2.8 ± 0.7
human_rate_mean = 4.0
human_rat_std = 1.2
rat_rate_mean = 2.8
rat_rate_std = 0.7

ind = np.arange(len(circuit_names))
width = 0.45

s1 = ax0.bar(ind, ratio_touches_edges_human, width, color='teal', label='Human')
s2 = ax0.bar(ind + width, ratio_touches_edges_rat, width, color='blueviolet',label='Rat')
ax0.set_xlim(ax0.get_xlim())
ax0.plot(ax0.get_xlim(), [4.0, 4.0], color="teal", ls="--", label='Hunt et al., 2022')
ax0.fill_between(ax0.get_xlim(), 4.0+1.2, 4.0-1.2, color='teal', alpha=0.3)
ax0.plot(ax0.get_xlim(), [2.8, 2.8], color="blueviolet", ls="--", label='Feldmeyer et al., 2006')
ax0.fill_between(ax0.get_xlim(), 2.8+0.7, 2.8-0.7, color='blueviolet', alpha=0.3)



ax0.set_xticks(range(len(circuit_names)), labels=circuit_names, fontsize=12,
              rotation=45, ha="right", rotation_mode="anchor")

ax0.set_title('Rate of touches over edges', fontsize=12)
ax0.set_ylabel('num. touches/num. edges', fontsize=12)
ax0.set_xlabel('Connectivity matrix', fontsize=12)
ax0.spines[['right', 'top']].set_visible(False)
ax0.legend(loc='center left', bbox_to_anchor=(1, 0.5))

#plt.savefig('Ratio_touches_edges.svg', bbox_inches='tight', transparent=True, dpi=100)
#plt.show()
```
