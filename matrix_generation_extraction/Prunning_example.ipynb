# Prunning methods

This is a notebook showing how to use different prunning methods on connectomes which are represented by sparse matrices with integer entries representing the number of touches/synapses in a connection. 

```python
import scipy.sparse as sp
import numpy as np 
from conntility import ConnectivityMatrix
from prunning import *
```

```python
# Load data 
root="/Users/natalibarroszulaica/Desktop/OBI/HUMAN_CORTEX/PAPERS_ON_GOING/CONNECTIVITY_PAPER/code/conn_matrix/"
fnames= {"structural":{"rat":
                       "Rat_623um_squared_struc_conmat_filtered_compressed.h5",
                       "human":
                       "Human_NEWstruct_conmat_filtered_compressed.h5"},
         "functional":{"rat":
                       "Rat_623um_squared_funct_conmat_filtered_compressed.h5",
                       "human":
                       "Human_NEWfunct_conmat_filtered_compressed.h5"}}

# Load matrices 
#animal="human"
animal="rat"

# Load functional matrices 
fname_func=fnames["functional"][animal]
A_functional=ConnectivityMatrix.from_h5(f"{root}{fname_func}").matrix.tocsr()

# Load structural data 
fname_struct=fnames["structural"][animal]
A_structural=ConnectivityMatrix.from_h5(f"{root}{fname_struct}").matrix.tocsr()

# For testing we use a small submatrix, comment out the next two lines to generate the full matrices
#A_structural=A_structural[np.ix_(range(100), range(100))] #just for testing  
#A_functional=func_mat[np.ix_(range(100), range(100))] #just for testing

#Target values from the functional matrices
target_edges=A_functional.nnz
target_touches=A_functional.sum()
```

### Choosing target values for edges and touches from the structural connectome

```python
print(f"Number of edges in structural {A_structural.nnz}")
print(f"Target number of edges {target_edges}")

print(f"\nNumber of touches in structural {A_structural.sum()}")
print(f"Target number of touches {target_touches}")
```

# Prunning to match edge count

Prunning use to reach a given graph density value i.e., a given global connection probability

```python
B=prune_edges_to_target_edges(A_structural, target_edges, 0)
print(f"Number of edges {B.nnz}")
print(f"Number of touches {B.sum()}")
```

```python
# We do this in an approximate way and the tolerance is chosen by the user
B=prune_touches_to_target_edges(A_structural, target_edges, seed=0, tol=0.01)
print(f"Number of edges {B.nnz}")
print(f"Number of touches {B.sum()}")
```

# Prunning to match touch count

Prunning use to reach a given number of synapses in a volume

```python
B=prune_touches_to_target_touches(A_structural, target_touches, seed=0)
print(f"Number of edges {B.nnz}")
print(f"Number of touches {B.sum()}")
```

```python
# We do this in an approximate way and the tolerance is chosen by the user
B=prune_edges_to_target_touches(A_structural, target_touches, seed=5, tol=0.01)
print(f"Number of edges {B.nnz}")
print(f"Number of touches {B.sum()}")
```
